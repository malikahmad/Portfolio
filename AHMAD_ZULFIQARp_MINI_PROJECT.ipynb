```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("/content/auto-mpg.csv")
```

```python
df.head()
```

```python
print(df.columns)
```

```python
# Number of rows and columns
print("Shape:", df.shape)

# Column names
print("\nColumns:")
print(df.columns)

# Dataset information
print("\nDataset Information:")
print(df.info())

# Statistical summary
print("\nStatistical Summary:")
print(df.describe())
```

```python
print("Missing values:")
print(df.isnull().sum())
```

```python
df.isnull().sum()
```

```python
df["horsepower"] = pd.to_numeric(df["horsepower"], errors="coerce")

print("Missing values after cleaning:")
print(df.isnull().sum())
```

```python
df = df.dropna()

print("Shape after removing missing values:", df.shape)
```

```python
print("Number of duplicate rows:", df.duplicated().sum())
```

```python
df = df.drop_duplicates()

print("Shape after removing duplicates:", df.shape)
```

```python
mean_mpg = np.mean(df["mpg"])

print("Mean MPG:", mean_mpg)
```

```python
min_mpg = np.min(df["mpg"])

print("Minimum MPG:", min_mpg)
```

```python
max_mpg = np.max(df["mpg"])

print("Maximum MPG:", max_mpg)
```

```python
print("Mean MPG:", df["mpg"].mean())
print("Minimum MPG:", df["mpg"].min())
print("Maximum MPG:", df["mpg"].max())
```

```python
high_mpg = df[df["mpg"] > 30]

print(high_mpg)
```

```python
print("Cars with MPG > 30:", len(high_mpg))
```

```python
high_hp = df[df["horsepower"] > 150]

print(high_hp)
```

```python
light_cars = df[df["weight"] < 2500]

print(light_cars)
```

```python
filtered_cars = df[
    (df["mpg"] > 30) &
    (df["horsepower"] < 100) &
    (df["weight"] < 2500)
]

print(filtered_cars)
```

```python
print("Number of matching cars:", len(filtered_cars))
```

```python
plt.figure(figsize=(8, 5))

plt.hist(df["mpg"], bins=20)

plt.title("Distribution of MPG")
plt.xlabel("Miles Per Gallon (MPG)")
plt.ylabel("Number of Cars")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(df["horsepower"], df["mpg"])

plt.title("MPG vs Horsepower")
plt.xlabel("Horsepower")
plt.ylabel("MPG")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(df["weight"], df["mpg"])

plt.title("MPG vs Car Weight")
plt.xlabel("Weight")
plt.ylabel("MPG")

plt.show()
```

```python
year_mpg = df.groupby("model year")["mpg"].mean()

plt.figure(figsize=(10, 5))

plt.plot(year_mpg.index, year_mpg.values, marker="o")

plt.title("Average MPG by Model Year")
plt.xlabel("Model Year")
plt.ylabel("Average MPG")

plt.show()
```

```python
cylinder_mpg = df.groupby("cylinders")["mpg"].mean()

plt.figure(figsize=(8, 5))

plt.bar(cylinder_mpg.index.astype(str), cylinder_mpg.values)

plt.title("Average MPG by Number of Cylinders")
plt.xlabel("Number of Cylinders")
plt.ylabel("Average MPG")

plt.show()
```

```python
total_cars = len(df)

cars_above_30 = len(df[df["mpg"] > 30])

probability = cars_above_30 / total_cars

print("Total cars:", total_cars)
print("Cars with MPG > 30:", cars_above_30)
print("Probability:", probability)
print("Probability (%):", probability * 100)
```

```python
probability = np.mean(df["mpg"] > 30)

print("Probability of selecting a car with MPG > 30:", probability)
print("Percentage:", probability * 100, "%")
```
